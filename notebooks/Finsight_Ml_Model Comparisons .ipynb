## Handiling differing Caplitization

```python
# Converting text to lowercase 

X_train["description_sa"] = X_train["description_sa"].str.lower()
X_val["description_sa"] = X_val["description_sa"].str.lower()
X_test["description_sa"] = X_test["description_sa"].str.lower()

print(X_train["description_sa"].head())
```

## Converting descriptions from categorical to numerical using TF-IDF

# Feature Combination C
#### Including all features except “split and seen_in_train”

### Dropping unecoisary columns 

```python

drop_columns = [
    "category",
    "category_raw",
    "split",
    "seen_in_train"
]

X_train_C = train.drop(columns=drop_columns).copy()
X_val_C = val.drop(columns=drop_columns).copy()
X_test_C = test.drop(columns=drop_columns).copy()

y_train_C = train["category"].copy()
y_val_C = val["category"].copy()
y_test_C = test["category"].copy()

print("Training features:", X_train_C.shape)
print("Validation features:", X_val_C.shape)
print("Testing features:", X_test_C.shape)

print("\nFeature columns:")
print(X_train_C.columns.tolist())
```

```python
print(X_train_C.dtypes)
```

### Prepare the text features

```python

text_columns = [
    "txn_date",
    "description_raw",
    "description_sa",
    "txn_type"
]

X_train_text_C = (
    X_train_C[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.lower()
    .str.split()
    .str.join(" ")
)

X_val_text_C = (
    X_val_C[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.lower()
    .str.split()
    .str.join(" ")
)

X_test_text_C = (
    X_test_C[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.lower()
    .str.split()
    .str.join(" ")
)

print(X_train_text_C.head())
```

### Conver the text into TF-IDF features

```python

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_C = TfidfVectorizer()

X_train_text_tfidf_C = tfidf_C.fit_transform(X_train_text_C)
X_val_text_tfidf_C = tfidf_C.transform(X_val_text_C)
X_test_text_tfidf_C = tfidf_C.transform(X_test_text_C)

print("Training:", X_train_text_tfidf_C.shape)
print("Validation:", X_val_text_tfidf_C.shape)
print("Testing:", X_test_text_tfidf_C.shape)
```

### Preparing the numeric features

```python

from sklearn.preprocessing import StandardScaler

numeric_columns_C = [
    "amount_usd",
    "amount_zar",
    "stmt_year",
    "stmt_month",
    "stmt_day",
    "purchase_year",
    "purchase_month",
    "purchase_day",
    "days_to_process",
    "has_purchase_date",
    "hour",
    "minute",
    "has_time",
    "is_online"
]

scaler_C = StandardScaler()

X_train_num_C = scaler_C.fit_transform(
    X_train_C[numeric_columns_C]
)

X_val_num_C = scaler_C.transform(
    X_val_C[numeric_columns_C]
)

X_test_num_C = scaler_C.transform(
    X_test_C[numeric_columns_C]
)

print("Training:", X_train_num_C.shape)
print("Validation:", X_val_num_C.shape)
print("Testing:", X_test_num_C.shape)
```

### Combine text and numeric features

```python

from scipy.sparse import hstack, csr_matrix

X_train_C_final = hstack([
    X_train_text_tfidf_C,
    csr_matrix(X_train_num_C)
]).tocsr()

X_val_C_final = hstack([
    X_val_text_tfidf_C,
    csr_matrix(X_val_num_C)
]).tocsr()

X_test_C_final = hstack([
    X_test_text_tfidf_C,
    csr_matrix(X_test_num_C)
]).tocsr()

print("Training:", X_train_C_final.shape)
print("Validation:", X_val_C_final.shape)
print("Testing:", X_test_C_final.shape)
```

# Feature Selection D 
Including only South-African Features

```python

text_features_D = [
    "description_sa",
    "txn_type"
]

numeric_features_D = [
    "amount_zar",
    "stmt_year",
    "stmt_month",
    "stmt_day",
    "purchase_year",
    "purchase_month",
    "purchase_day",
    "days_to_process",
    "has_purchase_date",
    "hour",
    "minute",
    "has_time",
    "is_online"
]

target = "category"
```

### Seperating Features and Target 

```python

X_train_D_text = train[text_features_D].fillna("").astype(str).agg(" ".join, axis=1)
X_val_D_text = val[text_features_D].fillna("").astype(str).agg(" ".join, axis=1)
X_test_D_text = test[text_features_D].fillna("").astype(str).agg(" ".join, axis=1)

X_train_D_num = train[numeric_features_D].copy()
X_val_D_num = val[numeric_features_D].copy()
X_test_D_num = test[numeric_features_D].copy()

y_train_D = train[target]
y_val_D = val[target]
y_test_D = test[target]

print("Training:", X_train_D_text.shape, X_train_D_num.shape)
print("Validation:", X_val_D_text.shape, X_val_D_num.shape)
print("Test:", X_test_D_text.shape, X_test_D_num.shape)
```

### Converting text into TF-IDF features 

```python

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_D = TfidfVectorizer(
    max_features=100000,
    ngram_range=(1, 2),
    min_df=2,
    sublinear_tf=True
)

X_train_D_tfidf = tfidf_D.fit_transform(X_train_D_text)
X_val_D_tfidf = tfidf_D.transform(X_val_D_text)
X_test_D_tfidf = tfidf_D.transform(X_test_D_text)

print("Training TF-IDF:", X_train_D_tfidf.shape)
print("Validation TF-IDF:", X_val_D_tfidf.shape)
print("Test TF-IDF:", X_test_D_tfidf.shape)
```

### Cpombining Numerical Features 

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from scipy.sparse import hstack, csr_matrix

# Handle missing numeric values
imputer_D = SimpleImputer(strategy="constant", fill_value=0)

X_train_D_num_imp = imputer_D.fit_transform(X_train_D_num)
X_val_D_num_imp = imputer_D.transform(X_val_D_num)
X_test_D_num_imp = imputer_D.transform(X_test_D_num)

# Scale numeric features
scaler_D = StandardScaler()

X_train_D_num_scaled = scaler_D.fit_transform(X_train_D_num_imp)
X_val_D_num_scaled = scaler_D.transform(X_val_D_num_imp)
X_test_D_num_scaled = scaler_D.transform(X_test_D_num_imp)

# Combine text and numeric features
X_train_D_final = hstack([
    X_train_D_tfidf,
    csr_matrix(X_train_D_num_scaled)
]).tocsr()

X_val_D_final = hstack([
    X_val_D_tfidf,
    csr_matrix(X_val_D_num_scaled)
]).tocsr()

X_test_D_final = hstack([
    X_test_D_tfidf,
    csr_matrix(X_test_D_num_scaled)
]).tocsr()

print("Training:", X_train_D_final.shape)
print("Validation:", X_val_D_final.shape)
print("Test:", X_test_D_final.shape)
```

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, accuracy_score

lr_model_D = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

lr_model_D.fit(X_train_D_final, y_train_D)

y_val_pred_lr_D = lr_model_D.predict(X_val_D_final)

print("Model D - Logistic Regression")
print("Validation Accuracy:", accuracy_score(y_val_D, y_val_pred_lr_D))
print("Validation Macro F1:", f1_score(y_val_D, y_val_pred_lr_D, average="macro"))
print("Validation Weighted F1:", f1_score(y_val_D, y_val_pred_lr_D, average="weighted"))
```
